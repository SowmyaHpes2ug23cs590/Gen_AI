```python
from transformers import pipeline
```

```python
ner = pipeline(
    "ner",
    model="dbmdz/bert-large-cased-finetuned-conll03-english",
    aggregation_strategy="simple"
)
```

```python
resume_text = """
Sowmya Hardageri is a third-year engineering student at PES University.
She has worked on projects related to Generative AI and NLP.
She completed an internship at CCNCS.
"""
```

```python
entities = ner(resume_text)

for entity in entities:
    print(entity)
```
