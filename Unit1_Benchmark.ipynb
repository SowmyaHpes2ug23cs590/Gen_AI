```python
# Unit 1 – Model Benchmarking Assignment
### Models: BERT vs RoBERTa vs BART
### Objective: Observe model behavior beyond intended architecture
```

```python
from transformers import pipeline
```

```python
#BERT Text Generation
try:
    bert_gen = pipeline("text-generation", model="bert-base-uncased")
    output = bert_gen("The future of Artificial Intelligence is", max_length=30)
    print(output)
except Exception as e:
    print("BERT Error:", e)
```

```python
#RoBERTa Text Generation
try:
    roberta_gen = pipeline("text-generation", model="roberta-base")
    output = roberta_gen("The future of Artificial Intelligence is", max_length=30)
    print(output)
except Exception as e:
    print("RoBERTa Error:", e)
```

```python
#BART Text Generation
bart_gen = pipeline("text-generation", model="facebook/bart-base")
output = bart_gen("The future of Artificial Intelligence is", max_length=30)
print(output)
```

```python
#BERT Fill-Mask
bert_mask = pipeline("fill-mask", model="bert-base-uncased")
bert_mask("The goal of Generative AI is to [MASK] new content.")
```

```python
#RoBERTa Fill Mask
roberta_mask = pipeline("fill-mask", model="roberta-base")
roberta_mask("The goal of Generative AI is to <mask> new content.")
```

```python
#BART Fill Mask
bart_mask = pipeline("fill-mask", model="facebook/bart-base")
bart_mask("The goal of Generative AI is to <mask> new content.")
```

```python
#BERT Question Answering
qa_bert = pipeline("question-answering", model="bert-base-uncased")

qa_bert(
    question="What are the risks?",
    context="Generative AI poses significant risks such as hallucinations, bias, and deepfakes."
)
```

```python
#RoBERTa QA
qa_roberta = pipeline("question-answering", model="roberta-base")

qa_roberta(
    question="What are the risks?",
    context="Generative AI poses significant risks such as hallucinations, bias, and deepfakes."
)
```

```python
#BART QA
qa_bart = pipeline("question-answering", model="facebook/bart-base")

qa_bart(
    question="What are the risks?",
    context="Generative AI poses significant risks such as hallucinations, bias, and deepfakes."
)
```
